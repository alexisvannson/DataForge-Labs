## Phase 1: Customer Segmentation

### Feature Engineering

We create meaningful features that capture customer behavior, value, and engagement patterns.

```python
# Calculate Age and Customer Tenure
df['Age'] = 2021 - df['Year_Birth']
df['Dt_Customer'] = pd.to_datetime(df['Dt_Customer'], format='%d-%m-%Y')
df['Customer_Tenure_Days'] = (pd.to_datetime('2021-01-01') - df['Dt_Customer']).dt.days

# Family size
df['Family_Size'] = df['Kidhome'] + df['Teenhome'] + 1
df['Has_Children'] = ((df['Kidhome'] + df['Teenhome']) > 0).astype(int)
df['Living_Alone'] = df['Marital_Status'].apply(lambda x: 1 if x in ['Single', 'Divorced', 'Widow', 'Alone', 'Absurd', 'YOLO'] else 0)

# Education level (simplified)
education_mapping = {'Basic': 1, '2n Cycle': 2, 'Graduation': 3, 'Master': 4, 'PhD': 5}
df['Education_Level'] = df['Education'].map(education_mapping)
```

```python
# Total Spending
df['Total_Spending'] = (df['MntWines'] + df['MntFruits'] + df['MntMeatProducts'] + df['MntFishProducts'] + df['MntSweetProducts'] + df['MntGoldProds'])

# Spending per category ratio
for category in ['MntWines', 'MntMeatProducts', 'MntFish Products', 'MntFruits', 'MntSweetProducts', 'MntGoldProds']:
    if category in df.columns:
        df[f'{category}_Ratio'] = df[category] / (df['Total_Spending'] + 1)
```

```python
# Campaign responsiveness score
df['Campaign_Response_Score'] = (df['AcceptedCmp1'] + df['AcceptedCmp2'] + df['AcceptedCmp3'] + df['AcceptedCmp4'] + df['AcceptedCmp5'])

# Total purchases and channel preferences
df['Total_Purchases'] = df['NumWebPurchases'] + df['NumCatalogPurchases'] + df['NumStorePurchases']

df['Web_Purchase_Ratio'] = df['NumWebPurchases'] / (df['Total_Purchases'] + 1)
df['Catalog_Purchase_Ratio'] = df['NumCatalogPurchases'] / (df['Total_Purchases'] + 1)
df['Store_Purchase_Ratio'] = df['NumStorePurchases'] / (df['Total_Purchases'] + 1)

# Average order value
df['Avg_Order_Value'] = df['Total_Spending'] / (df['Total_Purchases'] + 1)

df = df.dropna()
```

### Clustering Analysis

We use K-means clustering to identify distinct customer segments. The optimal number of clusters is determined using the Elbow method and Silhouette analysis.

```python
# For clustering, we typically want to exclude categorical variables (unless properly encoded), IDs, data leak features,
# and highly correlated/redundant variables. Here, we'll use all available numeric features except obvious identifiers or target-like columns.

exclude_cols = [
    'ID',
    'Response',             # Target
    'Z_CostContact',        # Constant
    'Z_Revenue',            # Constant
    'Dt_Customer',          # Date - already engineered tenure/recency
    'Education',            # Categorical raw
    'Marital_Status',       # Categorical raw

]

# Select all numeric columns except excluded
clustering_features = [
    col for col in df.select_dtypes(include=[np.number]).columns
    if col not in exclude_cols
]

# Prepare data for clustering
X_cluster = df[clustering_features].copy()

# Standardize features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_cluster)

print(f"✓ Prepared {X_scaled.shape[0]} customers with {X_scaled.shape[1]} features for clustering")
```

```python
# Determine optimal number of clusters
inertias = []
silhouette_scores = []
K_range = range(2, 11)

for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=20, max_iter=300)
    kmeans.fit(X_scaled)
    inertias.append(kmeans.inertia_)
    silhouette_scores.append(silhouette_score(X_scaled, kmeans.labels_))

# Visualization
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 5))

ax1.plot(K_range, inertias, 'bo-', linewidth=2, markersize=8)
ax1.set_xlabel('Number of Clusters', fontsize=12)
ax1.set_ylabel('Inertia', fontsize=12)
ax1.set_title('Elbow Method - Optimal K Selection', fontsize=14, fontweight='bold')
ax1.grid(True, alpha=0.3)

ax2.plot(K_range, silhouette_scores, 'ro-', linewidth=2, markersize=8)
ax2.set_xlabel('Number of Clusters', fontsize=12)
ax2.set_ylabel('Silhouette Score', fontsize=12)
ax2.set_title('Silhouette Score Analysis', fontsize=14, fontweight='bold')
ax2.grid(True, alpha=0.3)
ax2.axhline(y=0.5, color='g', linestyle='--', alpha=0.5, label='Good threshold (0.5)')
ax2.legend()

plt.tight_layout()
plt.show()

print("\nSilhouette Scores:")
for k, score in zip(K_range, silhouette_scores):
    print(f"  K={k}: {score:.4f}")

# Recommend optimal K
optimal_k = silhouette_scores.index(max(silhouette_scores[2:6])) + 2  # Between 4-8 clusters
print(f"\n✓ Recommended number of clusters: {optimal_k}")
```

```python
# Fit final clustering model
optimal_k = 4

kmeans_final = KMeans(n_clusters=optimal_k, random_state=42, n_init=50, max_iter=500)
df['Segment'] = kmeans_final.fit_predict(X_scaled)

print(f"Silhouette Score: {silhouette_score(X_scaled, df['Segment']):.4f}")
print(df['Segment'].value_counts().sort_index())
```

## Clustering Analysis Results

### K-Means Performance Assessment

After conducting K-Means clustering on the retail tabular data, the analysis reveals suboptimal results:

- **Silhouette Score**: 0.15
- **Method**: Combined elbow method and silhouette analysis

A silhouette score of 0.15 indicates weak cluster structure, suggesting that the identified clusters have significant overlap and are not well-separated. This value is considerably below the threshold for meaningful clustering (typically ≥ 0.5 for acceptable clusters), rendering the K-Means results insufficient for actionable segmentation.

### Next Steps: Alternative Clustering Approaches

Given the poor performance of K-Means, we will explore alternative clustering methodologies:

1. **DBSCAN (Density-Based Spatial Clustering of Applications with Noise)**
   - Better suited for identifying clusters of arbitrary shapes
   - Robust to outliers and noise in the dataset
   - Does not require pre-specification of cluster numbers


This alternative method will be evaluated to determine whether more meaningful cluster structures can be identified in the retail dataset.

# Let's find the best value for epsilon for the DBSCAN

```python
from sklearn.cluster import DBSCAN
from sklearn.neighbors import NearestNeighbors
import matplotlib.pyplot as plt
import numpy as np

# 1. reuse your existing X_scaled from the previous code
# Ensure X_scaled is defined (as per your previous snippet)

# 2. define min_samples
# Rule of thumb: 2 * n_features, or simply 5 or 10 for customer data
min_samples = 2 * X_scaled.shape[1]

# 3. Calculate k-nearest neighbors
# We look at the distance to the (min_samples)-th nearest neighbor
neighbors = NearestNeighbors(n_neighbors=min_samples)
neighbors_fit = neighbors.fit(X_scaled)
distances, indices = neighbors_fit.kneighbors(X_scaled)

# 4. Sort and plot distances to find the "Elbow"
distances = np.sort(distances[:, min_samples-1], axis=0)

plt.figure(figsize=(10, 6))
plt.plot(distances)
plt.title('K-distance Graph to find Optimal eps', fontsize=14, fontweight='bold')
plt.xlabel('Data Points sorted by distance', fontsize=12)
plt.ylabel(f'Distance to {min_samples}th Nearest Neighbor (Epsilon)', fontsize=12)
plt.grid(True, alpha=0.3)
plt.show()

print(f"✓ Look for the 'knee' (sharp bend) in the graph. The Y-axis value at that point is your optimal 'eps'.")
```

```python
from sklearn.cluster import DBSCAN
from sklearn.metrics import silhouette_score
from sklearn.decomposition import PCA
import pandas as pd

# --- PARAMETERS TO TUNE ---
CHOSEN_EPS = 7.5    # <--- REPLACE THIS based on the plot above (Y-axis value at the bend)
MIN_SAMPLES = 2 * X_scaled.shape[1]
# --------------------------

# 1. Fit DBSCAN
dbscan = DBSCAN(eps=CHOSEN_EPS, min_samples=MIN_SAMPLES)
labels = dbscan.fit_predict(X_scaled)

# 2. Analyze Results
# DBSCAN labels noise as -1. We usually want to count how much noise we have.
n_clusters_ = len(set(labels)) - (1 if -1 in labels else 0)
n_noise_ = list(labels).count(-1)

print(f"✓ DBSCAN found {n_clusters_} clusters and {n_noise_} noise points.")

# 3. Evaluate (Only if we have more than 1 cluster)
if n_clusters_ > 1:
    sil_score = silhouette_score(X_scaled, labels)
    print(f"✓ Silhouette Score: {sil_score:.4f}")
else:
    print("! Warning: Only 1 cluster found. Try decreasing 'eps' or 'min_samples'.")

# 4. Visualize with PCA (2D Projection)
# Since we have many features, we project to 2D to see the separation
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

plt.figure(figsize=(10, 8))
unique_labels = set(labels)
colors = [plt.cm.Spectral(each) for each in np.linspace(0, 1, len(unique_labels))]

for k, col in zip(unique_labels, colors):
    if k == -1:
        # Black used for noise.
        col = [0, 0, 0, 1]
        label_name = "Noise"
    else:
        label_name = f"Cluster {k}"

    class_member_mask = (labels == k)

    xy = X_pca[class_member_mask]
    plt.plot(xy[:, 0], xy[:, 1], 'o', markerfacecolor=tuple(col),
             markeredgecolor='k', markersize=6 if k == -1 else 10, label=label_name)

plt.title(f'DBSCAN Clustering (eps={CHOSEN_EPS})', fontsize=14)
plt.xlabel('PCA Component 1')
plt.ylabel('PCA Component 2')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

### DBSCAN Clustering – Interpretation and Business Implications

Using the optimal parameter value **ε = 7.5**, identified through the K-distance (elbow) method, the DBSCAN algorithm detected **a single cluster** and **63 observations classified as noise**.

#### Interpretation of the Result
The outcome reflects the **true density structure of the data rather than a parameter misspecification**. The fact that DBSCAN identifies only one cluster indicates that the customer population forms a **single, continuous density region**, with no clearly separable dense subgroups.

The noise points correspond to a limited number of customers exhibiting atypical behavior, lying in low-density areas of the feature space. These observations represent **outliers rather than meaningful customer segments**.

#### Business Interpretation for Retail Analytics
From a retail perspective, this result implies that the customer base **does not naturally segment into distinct behavioral groups** when viewed through a density-based lens. Instead, customers appear to form a relatively homogeneous population with continuous variation in purchasing and engagement patterns.

As a consequence:
- Density-based segmentation is **not appropriate** for this dataset.
- Marketing or personalization strategies should not rely on DBSCAN-style clustering to define customer groups.

#### Conclusion
With ε optimally chosen, DBSCAN provides a clear and informative result: **the retail customer dataset does not contain inherent density-based segments**. This finding is itself a valuable insight, guiding analysts away from unsuitable segmentation methods and toward more effective analytical strategies.

## Phase 2: Advanced Analytics

### 2.1 Campaign Response Prediction Model

Build a predictive model to identify which customers are most likely to respond to future campaigns.

```python
# Prepare data for campaign response prediction
# Target: Whether the customer responded to the last marketing campaign
# One-hot encode 'Education' and 'Marital_Status'
df = pd.get_dummies(df, columns=['Education', 'Marital_Status'], drop_first=True)
df = df.drop('Dt_Customer', axis=1)

# Select features: all numeric columns except 'Response'
numeric_cols = df.select_dtypes(include='number').columns
X = df[[col for col in numeric_cols if col != 'Response']]
y = df['Response']

# Train-test split (now X and y have the same number of rows)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

# Scale features for models sensitive to scale (Logistic Regression, SGDClassifier, SVM)
# Tree-based models are generally not sensitive to scaling, but it won't harm them.
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Convert scaled arrays back to DataFrame, preserving column names and index
X_train = pd.DataFrame(X_train_scaled, columns=X_train.columns, index=X_train.index)
X_test = pd.DataFrame(X_test_scaled, columns=X_test.columns, index=X_test.index)

print(f"Training set: {X_train.shape[0]} samples")
print(f"Test set: {X_test.shape[0]} samples")
print(f"Response rate in training: {y_train.mean():.1%}")
print(f"Response rate in test: {y_test.mean():.1%}")
```

```python
# Core ML libraries
from sklearn.linear_model import LogisticRegression, SGDClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC

# Metrics
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

# Experiment tracking and optimization
import mlflow
import mlflow.sklearn
import optuna

# Data manipulation
import pandas as pd
import numpy as np
def train_models_with_tuning(
    X_train,
    X_test,
    y_train,
    y_test,
    models_to_train=None,
    n_trials=10,
    threshold=0.35,
    experiment_name="Model_Selection",
    random_state=42
):
    """
    Train and tune multiple classification models using Optuna for hyperparameter optimization
    and MLflow for experiment tracking.
    """

    # Define all available models
    available_models = {
        'Logistic Regression': {},
        'Random Forest': {},
        'Gradient Boosting': {},
        'Ridge Classifier': {},
        'Naive Bayes': {},
        'Support Vector Machine': {}
    }

    # Filter models if specified
    if models_to_train is not None:
        models = {k: v for k, v in available_models.items() if k in models_to_train}
        if not models:
            raise ValueError(f"No valid models selected. Available: {list(available_models.keys())}")
    else:
        models = available_models

    def objective(trial, clf_name):
        """Optuna objective function for hyperparameter optimization."""
        with mlflow.start_run(nested=True, run_name=f"{clf_name}_trial"):
            if clf_name == 'Logistic Regression':
                C = trial.suggest_loguniform('C', 1e-3, 10)
                solver = trial.suggest_categorical('solver', ['liblinear', 'lbfgs'])
                model = LogisticRegression(random_state=random_state, max_iter=1000, C=C, solver=solver)
                mlflow.log_param('C', C)
                mlflow.log_param('solver', solver)

            elif clf_name == 'Random Forest':
                n_estimators = trial.suggest_int('n_estimators', 50, 200)
                max_depth = trial.suggest_int('max_depth', 3, 20)
                min_samples_split = trial.suggest_int('min_samples_split', 2, 10)
                model = RandomForestClassifier(
                    random_state=random_state,
                    n_estimators=n_estimators,
                    max_depth=max_depth,
                    min_samples_split=min_samples_split
                )
                mlflow.log_param('n_estimators', n_estimators)
                mlflow.log_param('max_depth', max_depth)
                mlflow.log_param('min_samples_split', min_samples_split)

            elif clf_name == 'Gradient Boosting':
                n_estimators = trial.suggest_int('n_estimators', 50, 200)
                max_depth = trial.suggest_int('max_depth', 2, 10)
                learning_rate = trial.suggest_loguniform('learning_rate', 0.01, 0.3)
                model = GradientBoostingClassifier(
                    random_state=random_state,
                    n_estimators=n_estimators,
                    max_depth=max_depth,
                    learning_rate=learning_rate
                )
                mlflow.log_param('n_estimators', n_estimators)
                mlflow.log_param('max_depth', max_depth)
                mlflow.log_param('learning_rate', learning_rate)

            elif clf_name == 'Ridge Classifier':
                alpha = trial.suggest_loguniform('alpha', 1e-4, 10)
                model = SGDClassifier(
                    loss='log_loss',
                    penalty='l2',
                    alpha=alpha,
                    random_state=random_state,
                    max_iter=1000,
                    tol=1e-3,
                    fit_intercept=True,
                    class_weight='balanced',
                    n_jobs=-1
                )
                mlflow.log_param('alpha', alpha)

            elif clf_name == 'Naive Bayes':
                var_smoothing = trial.suggest_loguniform('var_smoothing', 1e-10, 1e-5)
                model = GaussianNB(var_smoothing=var_smoothing)
                mlflow.log_param('var_smoothing', var_smoothing)

            elif clf_name == 'Support Vector Machine':
                C = trial.suggest_loguniform('C', 1e-3, 10)
                kernel = trial.suggest_categorical('kernel', ['linear', 'rbf'])
                gamma = trial.suggest_loguniform('gamma', 1e-4, 1) if kernel == 'rbf' else 'scale'
                model = SVC(probability=True, random_state=random_state, C=C, kernel=kernel, gamma=gamma)
                mlflow.log_param('C', C)
                mlflow.log_param('kernel', kernel)
                mlflow.log_param('gamma', gamma)
            else:
                raise ValueError(f"Unknown classifier: {clf_name}")

            model.fit(X_train, y_train)
            y_pred_proba = model.predict_proba(X_test)[:, 1]
            roc_auc = roc_auc_score(y_test, y_pred_proba)
            mlflow.log_metric('roc_auc', roc_auc)
            return roc_auc

    results = {}
    best_models = {}

    # Main MLflow run
    with mlflow.start_run(run_name=experiment_name):
        mlflow.log_param('n_trials', n_trials)
        mlflow.log_param('threshold', threshold)
        mlflow.log_param('random_state', random_state)

        for clf_name in models.keys():
            with mlflow.start_run(run_name=clf_name, nested=True):
                # Optuna optimization
                study = optuna.create_study(direction='maximize')
                study.optimize(
                    lambda trial: objective(trial, clf_name),
                    n_trials=n_trials,
                    show_progress_bar=False
                )

                best_params = study.best_params

                # Train final model with best parameters
                if clf_name == 'Logistic Regression':
                    model = LogisticRegression(random_state=random_state, max_iter=1000, **best_params)
                elif clf_name == 'Random Forest':
                    model = RandomForestClassifier(random_state=random_state, **best_params)
                elif clf_name == 'Gradient Boosting':
                    model = GradientBoostingClassifier(random_state=random_state, **best_params)
                elif clf_name == 'Ridge Classifier':
                    model = SGDClassifier(
                        loss='log_loss',
                        penalty='l2',
                        random_state=random_state,
                        max_iter=1000,
                        tol=1e-3,
                        fit_intercept=True,
                        class_weight='balanced',
                        n_jobs=-1,
                        **best_params
                    )
                elif clf_name == 'Naive Bayes':
                    model = GaussianNB(**best_params)
                elif clf_name == 'Support Vector Machine':
                    model = SVC(probability=True, random_state=random_state, **best_params)

                model.fit(X_train, y_train)

                # Predictions with custom threshold
                y_pred_proba = model.predict_proba(X_test)[:, 1]
                y_pred_custom = (y_pred_proba >= threshold).astype(int)

                # Calculate metrics
                acc = accuracy_score(y_test, y_pred_custom)
                prec = precision_score(y_test, y_pred_custom, zero_division=0)
                rec = recall_score(y_test, y_pred_custom, zero_division=0)
                f1 = f1_score(y_test, y_pred_custom, zero_division=0)
                roc_auc = roc_auc_score(y_test, y_pred_proba)

                results[clf_name] = {
                    'Accuracy': acc,
                    'Precision': prec,
                    'Recall': rec,
                    'F1 Score': f1,
                    'ROC AUC': roc_auc
                }
                best_models[clf_name] = model

                # Log to MLflow
                mlflow.log_params(best_params)
                mlflow.log_metric('accuracy', acc)
                mlflow.log_metric('precision', prec)
                mlflow.log_metric('recall', rec)
                mlflow.log_metric('f1_score', f1)
                mlflow.log_metric('roc_auc', roc_auc)
                mlflow.sklearn.log_model(model, f"{clf_name}_best_model")

        # Create results DataFrame
        results_df = pd.DataFrame(results).T
        print("Model results")
        print(results_df.round(4))

        # Identify and log best model
        best_model_name = results_df['ROC AUC'].idxmax()
        best_model = best_models[best_model_name]
        best_roc_auc = results_df.loc[best_model_name, 'ROC AUC']

        print(f"Best model: {best_model_name}")

        mlflow.log_param('best_model', best_model_name)
        mlflow.log_metric('best_roc_auc', best_roc_auc)
        mlflow.sklearn.log_model(best_model, f"{best_model_name}_production")

    return {
        'results_df': results_df,
        'best_model_name': best_model_name,
        'best_model': best_model,
        'best_models': best_models
    }
```

```python
# Train all models
results = train_models_with_tuning(X_train, X_test, y_train, y_test, n_trials=20)

# Train specific models only
results = train_models_with_tuning(
    X_train, X_test, y_train, y_test,
    models_to_train=['Random Forest', 'Gradient Boosting', 'Logistic Regression'],
    n_trials=15,
    threshold=0.4
)

# Access results
print(results['results_df'])
best_model = results['best_model']
predictions = best_model.predict(X_test)
```

```python
# Define churn based on Recency (independent of any segments)
df['Churn'] = (df['Recency'] > 45).astype(int)

print("=" * 60)
print("CHURN TARGET DISTRIBUTION")
print("=" * 60)
print(df['Churn'].value_counts())
print(f"\nChurn Rate: {df['Churn'].mean():.2%}")
print(f"Active Customers: {(df['Churn'] == 0).sum():,}")
print(f"Churned Customers: {(df['Churn'] == 1).sum():,}")
print("=" * 60)

# Visualize churn distribution
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Pie chart
churn_counts = df['Churn'].value_counts()
axes[0].pie(churn_counts, labels=['Active', 'Churned'], autopct='%1.1f%%',
            colors=['#2ecc71', '#e74c3c'], startangle=90)
axes[0].set_title('Churn Distribution', fontsize=14, fontweight='bold')

# Recency distribution by churn status
df[df['Churn'] == 0]['Recency'].hist(bins=30, alpha=0.7, label='Active',
                                       color='#2ecc71', ax=axes[1])
df[df['Churn'] == 1]['Recency'].hist(bins=30, alpha=0.7, label='Churned',
                                       color='#e74c3c', ax=axes[1])
axes[1].axvline(x=45, color='black', linestyle='--', linewidth=2, label='Churn Threshold (45 days)')
axes[1].set_xlabel('Recency (Days)', fontsize=12)
axes[1].set_ylabel('Number of Customers', fontsize=12)
axes[1].set_title('Recency Distribution by Churn Status', fontsize=14, fontweight='bold')
axes[1].legend()

plt.tight_layout()
plt.show()
```

## 2. Feature Engineering for Churn Prediction

### Features Used
We use **only behavioral and transactional features**:

**Purchase Behavior:**
- Total spending (MntWines, MntMeat, MntFish, MntSweets, MntFruits, MntGold)
- Purchase frequency across channels (NumWebPurchases, NumCatalogPurchases, NumStorePurchases)
- Recency of last purchase

**Engagement Metrics:**
- Website visits (NumWebVisitsMonth)
- Campaign acceptance history (AcceptedCmp1-5, Response)

**Customer Profile:**
- Tenure (Customer_Days)
- Income
- Household composition (Kidhome, Teenhome)
- Education level (one-hot encoded)
- Marital status (one-hot encoded)



```python
# Prepare features for churn prediction (NO SEGMENTS, NO RECENCY, NO LEAKAGE)
print("=" * 60)
print("PREPARING CHURN PREDICTION FEATURES")
print("=" * 60)

# One-hot encode categorical variables if not already done
if 'Education_Graduation' not in df.columns and 'Education' in df.columns:
    df = pd.get_dummies(df, columns=['Education', 'Marital_Status'], drop_first=True)

# Define features to EXCLUDE from churn prediction
exclude_for_churn = [
    # Identifiers and targets
    'ID',
    'Churn',                    # Target variable

    # Data leakage - these define or predict churn
    'Recency',                  # CRITICAL: This defines Churn!
    'Churn_Probability',        # Model prediction from previous run
    'Churn_Prediction',         #  Model prediction from previous run
    'Churn_Risk',               # Derived from Churn_Probability

    # Clustering results - meaningless (Silhouette = 0.15)
    'Segment',
    'Segment_Name',

    # Other model predictions
    'Response_Probability',     # From campaign response model

    # Constants
    'Z_CostContact',
    'Z_Revenue',

    # Dates (already engineered)
    'Dt_Customer',

    # Potentially leaky - highly correlated with Response/Recency
    'Response',                 # May correlate with active behavior
    'Campaign_Response_Score',  # May encode recent activity
]

# Select all numeric columns except excluded ones
churn_features = [
    col for col in df.select_dtypes(include=['number']).columns
    if col not in exclude_for_churn
]

print(f"\n✓ Total features for churn prediction: {len(churn_features)}")
print(f"\nFeatures used:")
for i, feature in enumerate(churn_features, 1):
    print(f"  {i:2d}. {feature}")

print(f"\n  EXCLUDED to prevent data leakage:")
excluded_present = [col for col in exclude_for_churn if col in df.columns]
for i, feature in enumerate(excluded_present, 1):
    print(f"  {i:2d}. {feature}")

# Create feature matrix and target
X = df[churn_features].copy()
y = df['Churn']

# Verify no leakage
print(f"\n Verification:")
print(f"  - 'Recency' in features: {'Recency' in X.columns}")
print(f"  - 'Churn_Probability' in features: {'Churn_Probability' in X.columns}")
print(f"  - 'Churn_Prediction' in features: {'Churn_Prediction' in X.columns}")
print(f"  - All should be False ✓")

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

# Standardize features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Convert back to DataFrame for interpretability
X_train = pd.DataFrame(X_train_scaled, columns=X_train.columns, index=X_train.index)
X_test = pd.DataFrame(X_test_scaled, columns=X_test.columns, index=X_test.index)

print("\n" + "=" * 60)
print("DATA SPLIT SUMMARY")
print("=" * 60)
print(f"Training set: {X_train.shape[0]:,} samples, {X_train.shape[1]} features")
print(f"Test set: {X_test.shape[0]:,} samples, {X_test.shape[1]} features")
print(f"\nChurn rate in training: {y_train.mean():.2%}")
print(f"Churn rate in test: {y_test.mean():.2%}")
print("=" * 60)
print("\n✓ Features prepared and scaled")
print("✓ NO DATA LEAKAGE: Recency, predictions, and derived features excluded")
```

## 3. Train Churn Prediction Models

We'll train multiple models using the existing `train_models_with_tuning()` function with Optuna hyperparameter optimization and MLflow tracking.

```python
# Train churn prediction models (using existing function)
print("=" * 60)
print("TRAINING CHURN PREDICTION MODELS")
print("=" * 60)
print("This may take several minutes...\n")

# Train with adjusted threshold for better recall
churn_results = train_models_with_tuning(
    X_train=X_train,
    X_test=X_test,
    y_train=y_train,
    y_test=y_test,
    models_to_train=['Logistic Regression', 'Random Forest', 'Gradient Boosting'],
    n_trials=20,
    threshold=0.4,
    experiment_name="Churn_Prediction",
    random_state=42
)

# Extract results
results_df = churn_results['results_df']
best_churn_model = churn_results['best_model']
best_churn_model_name = churn_results['best_model_name']

print("\n" + "=" * 60)
print(f"BEST MODEL: {best_churn_model_name}")
print("=" * 60)
```

## 4. Identify Key Churn Drivers

Understanding which features drive churn helps create targeted retention strategies.

```python
import eli5
from eli5.sklearn import PermutationImportance

print("=" * 60)
print("CHURN DRIVERS - FEATURE IMPORTANCE ANALYSIS")
print("=" * 60)

# For tree-based models, use built-in feature importance
if hasattr(best_churn_model, 'feature_importances_'):
    feature_importance = pd.DataFrame({
        'Feature': X.columns,
        'Importance': best_churn_model.feature_importances_
    }).sort_values('Importance', ascending=False)

    print(f"\nTop 15 Churn Drivers ({best_churn_model_name}):\n")
    print(feature_importance.head(15).to_string(index=False))

    # Visualize
    plt.figure(figsize=(12, 8))
    sns.barplot(x='Importance', y='Feature', data=feature_importance.head(15), palette='rocket')
    plt.title(f'Top 15 Churn Drivers ({best_churn_model_name})', fontsize=16, fontweight='bold')
    plt.xlabel('Importance Score', fontsize=12)
    plt.ylabel('Feature', fontsize=12)
    plt.tight_layout()
    plt.show()

# For non-tree models (SVM, Logistic Regression), use permutation importance
else:
    print(f"\n{best_churn_model_name} doesn't have built-in feature importance.")
    print("Calculating permutation importance (this may take a minute)...\n")

    perm = PermutationImportance(best_churn_model, random_state=42, n_iter=10)
    perm.fit(X_test, y_test)

    # Get feature importance
    feature_importance = pd.DataFrame({
        'Feature': X.columns,
        'Importance': perm.feature_importances_
    }).sort_values('Importance', ascending=False)

    print("Top 15 Churn Drivers (Permutation Importance):\n")
    print(feature_importance.head(15).to_string(index=False))

    # Visualize
    plt.figure(figsize=(12, 8))
    sns.barplot(x='Importance', y='Feature', data=feature_importance.head(15), palette='rocket')
    plt.title('Top 15 Churn Drivers (Permutation Importance)', fontsize=16, fontweight='bold')
    plt.xlabel('Importance Score', fontsize=12)
    plt.ylabel('Feature', fontsize=12)
    plt.tight_layout()
    plt.show()

print("\n" + "=" * 60)
```
